```python
import pandas as pd
import numpy as np
import duckdb
import altair as alt
```

```python
# most popular top 10
most_popular_df = pd.read_csv(r'data/most-popular.tsv', sep='\t')

# all week global
all_weeks_global_df = pd.read_csv(r'data/all-weeks-global.tsv', sep='\t')

# all week countries
all_weeks_countries_df = pd.read_csv(r'data/all-weeks-countries.tsv', sep='\t')
```

```python
alt.Chart(most_popular_df).mark_bar().encode(
    alt.X('hours_viewed_first_91_days:Q', bin=alt.Bin(maxbins=20)),
    y='count()',
    color='category',
    tooltip=['hours_viewed_first_91_days', 'category', 'count()']
).properties(
    title='Histogram of Hours Viewed in First 91 Days'
).interactive()
```
